```python
import pandas as pd
import re

df = pd.read_csv('./가전 구독/LG_subscription.csv')

# 1. 너무 짧은 본문 제거 (100자 미만)
df = df[df['full_text'].str.len() >= 100]

# 2. 수집 실패 데이터 제거
df = df[~df['full_text'].str.contains('본문 추출 실패|본문 내용 없음|회원 공개글', na=False)]

# 3. 광고/이벤트성 노이즈 제거
noise_keywords = ['응모', '경품', '이벤트 참여', '댓글 이벤트', '협찬', '렌탈샵','업체',
                  '반도체', '금융', '로봇', '채용', '기술직','지원','원고료',]
pattern = '|'.join(noise_keywords)
df = df[~df['full_text'].str.contains(pattern, na=False)]

# 4. 가전 구독 관련 키워드 포함된 글만 유지
must_keywords = ['구독', '렌탈', '가전', 'LG', '엘지']
pattern2 = '|'.join(must_keywords)
df = df[df['full_text'].str.contains(pattern2, na=False)]

df = df.reset_index(drop=True)
print(f'필터링 후 데이터: {len(df)}건')
df.to_csv('./LG_subscription_filtered.csv', index=False, encoding='utf-8-sig')
```
