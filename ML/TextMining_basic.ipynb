-----
#### 3. 토큰화
- 텍스트를 의미있는 최소단위(token)으로 분리하는 과정
- 특정 기준을 가지고 분리 (문단, 문장, 단어, 형태소)
- 한국어는 교착어형태로서 주로 형태소 분석을 진행
    - 형태소 분석기
        - konlpy (mecab,kkma,okt,komoran,hannanum)
        - kiwi

#### 4. 수치화
- 텍스트 토큰을 숫자형태로 변환
- 단순 수치화 , 의미있는 벡터화해주는 과정
    - 원핫인코딩
        - 단점 : 희소행렬 (sparse matrix)
    - BoW(Bag of Words)
        - CounterVectorize : 단순 카운트 기반
        - TF-IDF : 모든 문서에 자주나오는 단어는 패널티, 하나의 문서에 자주나오는 단어는 가중치
        - 단점 : 의미적 유사성을 반영하지 못함
    - 워드임베딩 (Word Embedding)
        - 의미적 유사성을 반영하여, 벡터화
        - Word2Vec
