```python
from langchain_core.prompts import PromptTemplate, ChatPromptTemplate
from langchain_openai import ChatOpenAI
from langchain_core.output_parsers import StrOutputParser
```

- 간단한 챗봇 만들기

```python
## llm 모델의 객체 생성 (gpt-4o-mini,)
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model = 'gpt-4o-mini',temperature = 0,max_tokens = 500)
```

```python
## 사주를 출력해주는 챗봇 (시스템 메시지와 휴먼메시지를 활용하여 프롬프트 구성)
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages([
    ('system','당신은 사주팔자를 분석하여 알려주는 사주 전문가입니다'),
    ('human','{birth}와 {time}를 통해 사주를 분석해줘')
])
## 체인연결

chain = prompt | llm | StrOutputParser()
```

```python
result = chain.invoke({'birth':'양력 1997-06-29','time':'오후 7시'})
```

```python
print(result)
```

```python
while True:
    human_input = input("입력 : ")

    if human_input == "end":
        break
    print(chain.invoke({'ymd':human_input}))
```

### 제로샷 프롬프팅 (Zero-Shot Prompting)

- zero-shot prompting : AI 모델이 사전 지식없이 새로운 답변을 출력하는 방법, 예시없이 단순 답변을 출력하는 방법
- 특징
    - 모델은 오로지 프롬프트 템플릿에 포함된 정보만을 활용하여 답변을 생성
    - 장점 : 빠르고, 간단하게 적용이 가능함
    - 단점 : 복잡한 문제나 구체적인 형식을 요구하는 작업에서는 모델의 응답이 기대에 못 미칠 수 있음

### Zero-Shot Prompting 개선방법
- Few-shot Prompting (예시를 제공)
- memory 기능 사용 (피드백을 통한 개선된 응답)
- RAG (검색 후 생성)

### 퓨샷 프롬프팅 (Few-Shot Prompting)
- 몇가지 예시를 프롬프트에 포함시켜 모델에게 원하는 출력형식이나 답변을 학습시키는 방식
- 특징
  - 예시를 제공하여 모델이 학습과 답변에 대한 안내 제공
  - 응답의 일관성과 정확도를 향상
  - 장점: 구체적으로 모델이 수행하여 출력해야 하는 내용을 인지시켜 원하는 결과를 얻을 수 있다
  - 단점: 적절한 예시를 선택하지 못하면 잘못된 결과를 얻을 수 있으므로 명확한 예시제공이 필요, 프롬프트가 길어질 수 있다.

```python
from langchain_core.prompts import FewShotPromptTemplate
```

```python
# 예시
# 입력값에 대한 반대말을 출력하는 예시
examples1 = [
    {"input": "happy", "output": "sad"},
    {"input": "tall", "output": "short"},
    {"input": "sunny", "output": "rainy"},
    {"input": "surprised", "output": "calm"},
    {"input": "dry", "output": "humid"},
    {"input": "hot", "output": "cold"},
    {"input": "satisfied", "output": "dissatisfied"}
]
```

```python
# 예시데이터 표현을 위한 기본 프롬프트 템플릿 생성
prompt = PromptTemplate.from_template('입력 : {input}\n 출력 : {output}')
```

```python
# fewshot Prompt template 생성
fewshot_prompt = FewShotPromptTemplate(
    examples = examples1,
    example_prompt = prompt,
    input_variables = ['input'], # 최종 사용자에게 입력받는 변수명
    # 전체 프롬프트 앞쪽에 붙는 추가 설명
    prefix = (
        '''
         입력의 반대말을 출력하고,
         일상생활에서 사용하는 반대말로 출력하고,
        사전에 없다면 "알 수 없음" 이라고 출력하세요.
        '''
    ),
    suffix = '입력 : {input}'
)
```

```python
## chain 연결
chain2 = fewshot_prompt | llm | StrOutputParser()
```

```python
chain2.invoke({'input':'최석진'})
```

```python
chain2.invoke({'input':'대한민국'})
```

### 해당 키워드에 맞게 단계별로 답변을 구성하는 퓨샷 프롬프팅 

```python
# 상품명을 입력받아 홈쇼핑 대본 자동으로 생성 서비스
examples2 = [
    {
        "question": "상품명: '에어프라이어', 특징: '간편 조리, 1300W 출력', 가격: '120,000원'일 때 단계별 대본 가이드 예시를 보여주세요.",
        "answer": (
            "1) 인사 및 상품 소개: \"안녕하세요 여러분! 오늘은 집에서 간편하게 튀김 요리를 할 수 있는 '에어프라이어'를 소개합니다!\"\n"
            "2) 주요 특징 설명: \"1300W의 강력한 출력으로 바삭함을 살려주고, 조리 시간이 단축됩니다.\"\n"
            "3) 사용 시나리오 제시: \"감자튀김, 치킨너겟, 채소구이까지 모두 간편하게 조리해보세요.\"\n"
            "4) 가격 및 혜택 안내: \"지금 구매하시면 120,000원에 무료 배송과 1년 무상 A/S가 제공됩니다.\"\n"
            "5) 콜 투 액션: \"아래 구매 버튼을 눌러 할인 혜택을 놓치지 마세요!\""
        )
    },
    {
        "question": "상품명: '블루투스 헤드폰', 특징: '노이즈 캔슬링, 20시간 배터리', 가격: '80,000원'일 때 단계별 대본 가이드 예시를 보여주세요.",
        "answer": (
            "1) 인사 및 관심 유도: \"안녕하세요! 오늘은 음악 감상에 최적화된 '블루투스 헤드폰'을 소개드립니다.\"\n"
            "2) 기능 강조: \"능동형 노이즈 캔슬링으로 외부 소음을 완벽 차단, 최대 20시간 배터리로 하루 종일 사용 가능해요.\"\n"
            "3) 데모 시연 제안: \"실제 사용해볼까요? 소음 많은 카페에서도 깨끗한 사운드를 경험해보세요.\"\n"
            "4) 특별 할인 안내: \"오늘 한정 5% 할인 쿠폰 코드: LIVE5 적용하세요!\"\n"
            "5) 구매 유도 문구: \"지금 바로 구매 버튼을 클릭하고, 편안한 음악 시간을 즐겨보세요!\""
        )
    }
]
```

```python
shop_prompt = PromptTemplate.from_template('입력 : {question} 출력 : {answer}')
    
ex = {"상품명: '무선 청소기', 특징: '1회충전 1시간 연속사용, 강력흡입', 가격: '1,000,000원'일 때 단계별 대본 가이드 예시를 보여주세요."}
```

```python
shopping_fewshot_prompt = FewShotPromptTemplate(
    examples = examples2,
    example_prompt = shop_prompt,
    input_variables = ['input'],
    prefix = (
        '''
        상품 소개, 관심 유도 시 소비자를 끌어들일만한 매력적인 포인트 자극적으로 강조하세요.
        기능강조, 시연 시 제품의 특징이 잘 부각되도록 작성하세요. 또한 제품의 장점을 더 매력적이고 구체적으로 어필하세요.
        사용 시나리오 제시 시, 매우 구체적인 사용 사례를 통해 제시하세요
        특별할인 및 혜택 안내 시, 소비자를 위한 할인혜택을 제공하면서도, 너무 과하지 않은 혜택을 제공하세요.
        마지막 콜투 액션, 구매유도 문구에서 소비자를 자극할만한 창의적인 멘트로 마지막으로 소비자의 구매를 유도하세요.
        
        '''
    ),
    suffix = '입력 : {input}'
)
```

```python
shop_chain = shopping_fewshot_prompt | llm | StrOutputParser()
```

```python
print(shop_chain.invoke(ex))
```
