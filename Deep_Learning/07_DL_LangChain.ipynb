### LangChain
- 대규모 언어모델(LLM)을 활용하여 어플리케이션 개발을 돕는 라이브러리

- LLM을 통하여 데이터를 분석하고 질문하거나 답을 주는 작업을 진행
- 주요 기능 
    - 문맥인식을 통한 추론 (제공된 문맥에 기반하여 어떤 대답을 해야하는지에 결과)

### 대규모 언어모델 LLM의 한계
- 일반적인 상황 (모델이 학습한 상황)에서의 응답능력은 탁월하나, 훈련한 적이 없는 영역에서는 성능이 떨어짐 -> 환각(Hallucination)
- LLM모델은 상태를 저장하지 않으므로 이전 대화내용을 기억하지 못함
- 서로 다른 시나리오(task)에 특화된 다양한 모델을 사용해야함

### 프롬프트 작성 원칙
- 프롬프트: LLM 모델에게 입력하는 명령어
- 모델이 최대한 정확하고 유용한 정보를 제공할 수 있도록 효과적인 프롬프트를 작성하는것이 매우 중요
1. 명확성, 구체성
  - 질문은 명확하고 구체적이여야한다.
2. 배경 정보를 포함
  - 모델이 문맥을 이해할 수 있도록 필요한 배경정보를 제공 (환각을 줄일 수 있음)
3. 간결성
  - 핵심 정보에 초점을 맞추고, 불필요한 정보 배제
4. 페르소나 설정
  - llm 에 역할 부여
  - 예시: 당신은 인공지능 전문가입니다.
5. 문체, 언어
  - 대화의 맥락에 맞는 언어와 문제를 선택
6. 예시 작성
  - 명확한 출력결과를 위하여 예시 제공

```python
# 프롬프트 템플릿 지정
from langchain_core.prompts import PromptTemplate,ChatPromptTemplate
```

- 프롬프트 템플릿
- 프롬프트(LLM에 입력하는 명령어)를 미리 구조화하여 간편하게 재사용할 수 있게 도와주는 방법
- 입력받는 값과 고정된 텍스트를 묶어 유연하게 프롬프트를 생성할 수 있도록 도와주는 역할

### LCEL (LangChain Expression)
### 파이프라인(|)으로 체인 연결
- '|' 기호를 사용하여 체인을 연결
  - prompt | LLM 모델 | 출력파서

```python
# 수도를 물어보는 템플릿을 생성
prompt = PromptTemplate.from_template('{country}의 수도는 어디야?')
# 사용자에게 country 즉, 나라이름만 입력받으면 해당 수도를 출력하는 템플릿 샏성
```

```python
# LLM 모델
llm = ChatOpenAI(model = 'gpt-4o-mini',temperature = 0, max_tokens = 500)
```

```python
# 체인연결
# 프롬프트템플릿 - llm - 
chain = prompt | llm
```

```python
# chain 구동
chain.invoke({'country':'대한민국'}).content
```

### ChatPromptTemplate
- 대화형 메시지를 입력으로 사용하고 대화형 메시지를 출력
- 대화의 맥락을 포함한 메시지를 처리할 때 사용하는 템플릿
- 대화의 연속성을 고려하여 응답메시지를 생성
- 역할 (role), 내용 (content)으로 구성

- 메시지의 역할과 내용을 기반으로 프롬프트를 구성 (from_messages())
- system : 시스템의 기능을 설명 (role)
- human : 사용자의 질문 미리 정의

```python
# ChatPromptTemplate
chat_prompt = ChatPromptTemplate.from_messages([
    ('system','이 시스템은 여행전문가입니다.'),
    ('human','{text}')
])
# text를 사용자에게 입력받으면 여행관련 지식을 추천해줌 (지역이름만 넣어도 추천)
```

```python
# 모델 확인
llm
```

```python
# 체인 생성
travel_Chain = chat_prompt | llm
```

```python
# 체인 구동
travel_Chain.invoke({'text':'스위스 대표 여행지 추천해줘'}).content
```

- AI 메시지를 활용하여 출력양식 지정하기

```python
# 출력형식 지정하기
prompt_template = ChatPromptTemplate.from_messages([
    ('system', "너는 여행전문가이고 여행지의 맛있는 음식에 대해 전문적으로 대답할 수 있어"),
    ('human', '{question}'),
    ('ai',""" 아래의 형식처럼 답변을 작성해주세요
    ======================
    타이틀
    ======================
    - 응답 1
      - 설명
    - 응답 2
      - 설명
    - 응답 3
      - 설명
    """)
])
```

```python
# 체인 연결
travel_chain_2 = prompt_template | llm
```

```python
# 체인 구동 -> 사용자에게 값을 입력받아 출력
human_input = input("여행지를 입력해주세요. 예시 : 광주광역시")
result = travel_chain_2.invoke({'question':human_input}).content
print(result)
```
