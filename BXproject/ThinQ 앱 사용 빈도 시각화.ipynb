```python
import matplotlib.pyplot as plt

# 데이터
categories = ['Inactive', 'Light', 'Middle', 'Heavy']
values = [80.4, 7.3, 6.9, 5.5]
colors = ['#4C6EF5', '#ADB5BD', '#ADB5BD', '#ADB5BD']

# 그래프 생성
plt.figure(figsize=(9, 5))

for cat, val, col in zip(categories, values, colors):
    bar = plt.bar(
        cat,
        val,
        color=col,
        label=cat
    )

    # 🔑 퍼센트 텍스트 표시
    plt.text(
        bar[0].get_x() + bar[0].get_width() / 2,
        bar[0].get_height() + 1,
        f'{val}%',
        ha='center',
        va='bottom',
        fontsize=10
    )

# 제목 및 축
plt.title('LG ThinQ 사용자 사용시간 기준 유저 분포', fontsize=13)
plt.ylabel('유저 비율 (%)')
plt.ylim(0, 90)

# 범례 (우측)
plt.legend(
    [
        'Inactive: 사용 이력 없음',
        'Light: 0.1시간 미만',
        'Middle: 0.1~0.3시간',
        'Heavy: 0.3시간 이상'
    ],
    title='관여도 기준',
    loc='center left',
    bbox_to_anchor=(1.02, 0.5)
)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

categories = ['Inactive', 'Light', 'Middle', 'Heavy']
values = [80.4, 7.3, 6.9, 5.5]

colors = [
    '#a50034',  # Inactive
    '#1F2937',  # Light
    '#9CA3AF',  # Middle
    '#F0ECE4'   # Heavy
]

plt.figure(figsize=(7, 7))

# 🔑 pie 반환값 받기
wedges, texts, autotexts = plt.pie(
    values,
    colors=colors,
    autopct='%1.1f%%',
    startangle=90,
    counterclock=False,
    wedgeprops={
        'edgecolor': 'white',
        'linewidth': 1.5
    }
)

# 🔥 내부 % 숫자 강조
for i, autotext in enumerate(autotexts):
    autotext.set_fontsize(14)      # 크기 키우기
    autotext.set_fontweight('bold')

    # Inactive, Light 만 흰색
    if i in [0, 1]:
        autotext.set_color('white')
    else:
        autotext.set_color('#212529')

# 제목
plt.title('LG ThinQ 사용자 사용시간 기준 유저 분포', fontsize=13, pad=25)

# 범례
legend_elements = [
    Patch(facecolor='#a50034', label='Inactive : 사용 이력 없음'),
    Patch(facecolor='#1F2937', label='Light : 0.1시간 미만'),
    Patch(facecolor='#9CA3AF', label='Middle : 0.1 ~ 0.3시간'),
    Patch(facecolor='#F0ECE4', label='Heavy : 0.3시간 이상')
]

plt.legend(
    handles=legend_elements,
    loc='upper center',
    bbox_to_anchor=(0.5, 1.02),
    ncol=2,
    frameon=False,
    fontsize=10
)

plt.tight_layout()
plt.show()
```

```python
#데이터
x = np.array([
    '2025.01','2025.02','2025.03','2025.04','2025.05','2025.06',
    '2025.07','2025.08','2025.09','2025.10','2025.11','2025.12'
])

y1 = np.array([757,696,770,759,799,832,856,777,712,682,594,595])  # SmartThings
y2 = np.array([119,119,124,128,144,166,194,191,179,156,140,145])  # LG ThinQ

# 그래프 생성
plt.figure(figsize=(9,5))

plt.plot(x, y1, marker='o', linewidth=2, label='SmartThings',color = '#0c4da2')
plt.plot(x, y2, marker='o', linewidth=2, label='LG ThinQ',color = '#a50034' )

# 각 점 위에 숫자 표시
for i in range(len(x)):
    plt.text(x[i], y1[i] + 15, f'{y1[i]}', ha='center', fontsize=9)
    plt.text(x[i], y2[i] + 15, f'{y2[i]}', ha='center', fontsize=9)

# y축 설정 (300 단위)
plt.ylim(0, 1000)
plt.yticks(np.arange(0, 901, 300))

# x축 설정 (월별 시간 단위)
plt.xticks(x, rotation=45)

# 격자(Grid) 설정
plt.grid(
    which='major',
    axis='both',
    linestyle='--',
    linewidth=0.7,
    alpha=0.5
)

# 제목 및 라벨
plt.title('최근 1년간 LG ThinQ vs SmartThings 월 사용자 추이')
plt.ylabel('사용자 수 (만 명)')
plt.xlabel('월 (2025년 기준)')

plt.legend()
plt.tight_layout()
plt.savefig('thinq_vs_smartthings_trend_grid.png', dpi=300)
plt.show()
```

```python
# 1. 데이터 설정
months = ['2025.07', '2025.08', '2025.09', '2025.10', '2025.11', '2025.12', '2026.01\n~ 2026.03']
# 이미지 기반 추정치 및 확정치(35.8)
churn_rates = [24.5, 29.0, 28.5, 38.0, 37.5, 31.5, 35.8] 

# 색상 설정
colors = ['#D3D3D3'] * 6 + ['#a50034'] # 앞의 6개는 회색, 마지막은 파란색

# 2. 그래프 생성
fig, ax = plt.subplots(figsize=(10, 6))

# 막대 그래프 그리기
bars = ax.bar(months, churn_rates, color=colors, width=0.5, zorder=3)

# 3. 스타일 디테일 작업

# 마지막 예측 구간 배경 강조 (axvspan)
ax.axvspan(5.5, 6.5, facecolor='#E8F0FF', alpha=0.5, zorder=1)

# Y축 범위 및 그리드 설정
ax.set_ylim(0, 45)
ax.set_yticks([0, 10, 20, 30, 40])
ax.grid(axis='y', linestyle='-', alpha=0.3, zorder=0)

# 테두리 제거 (Spines)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.spines['bottom'].set_color('#CCCCCC')

# X/Y축 라벨 설정
ax.set_ylabel('예상 이탈률 (%)', fontsize=10, color='#666666', labelpad=15)
ax.tick_params(axis='x', colors='#666666', labelsize=9)
ax.tick_params(axis='y') # Y축 눈금선 제거

# 마지막 막대 위에 텍스트 추가
ax.text(6, 37, '예상 이탈률\n35.8%', ha='center', va='bottom', 
        color='#a50034', fontweight='bold', fontsize=11)

# 4. 마무리
plt.tight_layout()
plt.show()
```

```python

months = ['2025.07', '2025.08', '2025.09', '2025.10', '2025.11', '2025.12', '2026.01\n~ 2026.03']
# 이미지상의 수치를 기반으로 한 데이터
churn_rates = [24.5, 29.0, 28.5, 38.0, 37.5, 31.5, 35.8] 

# 색상 설정 (과거: 회색, 미래: 파란색)
colors = ['#BDBDBD'] * 6 + ['#a50034']

# 2. 그래프 생성
fig, ax = plt.subplots(figsize=(10, 6))

# 막대 그래프 (둥근 모서리 대신 직선 막대로 통일성 강조)
bars = ax.bar(months, churn_rates, color=colors, width=0.5, zorder=3)

# 3. 스타일 적용 (테두리 및 그리드)
# 사방 테두리 유지
for spine in ax.spines.values():
    spine.set_visible(True)
    spine.set_color('#333333')

# 점선 그리드 설정 (x, y축 모두)
ax.grid(True, linestyle='--', alpha=0.5, zorder=0)

# 마지막 구간 강조 배경
# ax.axvspan(5.5, 6.5, facecolor='#a50034', alpha=0.4, zorder=1)

# 4. 세부 설정
ax.set_ylim(0, 45)
ax.set_yticks([0, 10, 20, 30, 40])
ax.set_ylabel('예상 이탈률 (%)', fontsize=10, fontweight='bold')
ax.set_title('이탈 추이로 보는 LG ThinQ 이탈 고객 전망', fontsize=14, pad=20, fontweight='bold')

# 막대 위에 수치 표시 (마지막 막대 강조)
for i, bar in enumerate(bars):
    height = bar.get_height()
    if i == len(bars) - 1: # 마지막 데이터
        ax.text(bar.get_x() + bar.get_width()/2., height + 1, 
                f'예상 이탈률\n{height}%', ha='center', va='bottom', 
                color='#a50034', fontweight='bold', fontsize=11)
       

# X축 라벨 회전 (필요 시)
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```
